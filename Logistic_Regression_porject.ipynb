```python
#we gone a remove multiple columns becouse no nee and thye give same value like adult_male,embark_town,alive,deck(very high null value),class,who
```

```python
df.drop(['deck','adult_male','embark_town','adult_male','class', 'who','alive'],axis=1,inplace=True)
df.info()
```

```python
#age also have null value so fill by mean
df['age']=df['age'].fillna(df['age'].mean())
df.info()
```

```python
df.isnull().sum()
```

```python
#embarked is string so we remove 2 null row
df.dropna(subset=['embarked'], inplace=True)
```

```python
def ploting(var,num):
    plt.subplot(2, 2, num)
    sns.histplot(df[var], kde=True, bins=30)
    plt.title(f'Distribution of {var}')
    plt.xlabel(f'number of {var}')
    plt.ylabel(f'number of people')
ploting('age', 1 )
ploting('fare', 2)
plt.tight_layout()
plt.show()
```
